define all cat_columns

```python
# Define column lists excluding the specific target for each model
cat_cols_p = ['Ticket Channel', 'Ticket Type', 'Ticket Subject', 'Ticket Description', 'Ticket Status', 'Resolution']
cat_cols_t = ['Ticket Channel', 'Ticket Subject', 'Ticket Description', 'Ticket Status', 'Resolution', 'Ticket Priority']
```

Text cleaning

convert text to Lower case ,
Remove punctuations ,
Apply tokenization

```python
import string
import nltk
nltk.download('punkt_tab')
from nltk.tokenize import word_tokenize
translator=str.maketrans('','',string.punctuation)
for cat in cat_cols_t:
  df[cat]=df[cat].astype(str).str.lower()
  df[cat]=df[cat].str.translate(translator)
  df[cat]=df[cat].apply(word_tokenize)
df.head()
```

Remove Stopwords

```python
import nltk
from nltk.corpus import stopwords
nltk.download('stopwords')
stop_words=set(stopwords.words('english'))
for cat in cat_cols_t:
  df[cat]=df[cat].apply(lambda x:[word for word in x if word not in stop_words ])
df.head()
```

applying lemataizaton

```python
from nltk.stem import WordNetLemmatizer
from nltk.corpus import wordnet

# Download necessary components
nltk.download('wordnet')
nltk.download('averaged_perceptron_tagger_eng')

lemmatizer = WordNetLemmatizer()

# Helper function to map NLTK POS tags to WordNet tags
def get_wordnet_pos(word):
    tag = nltk.pos_tag([word])[0][1][0].upper()
    tag_dict = {"J": wordnet.ADJ, "N": wordnet.NOUN, "V": wordnet.VERB, "R": wordnet.ADV}
    return tag_dict.get(tag, wordnet.NOUN)

# Apply lemmatization with POS tagging
for cat in cat_cols_t:
    df[cat] = df[cat].apply(lambda x: [lemmatizer.lemmatize(word, get_wordnet_pos(word)) for word in x])
df.tail()
```

remove nan values in the data

```python
# Replace empty lists with a list containing 'none'
for cat in cat_cols_t:
    df[cat] = df[cat].apply(lambda x: ['none'] if len(x) == 0 else x)
df.head()
```
